```python
# ============================================================
# Reload All Tourism Datasets for Cleaning
# ============================================================

user_clean = pd.read_excel(os.path.join(data_path, "User.xlsx"))
city_clean = pd.read_excel(os.path.join(data_path, "City.xlsx"))
country_clean = pd.read_excel(os.path.join(data_path, "Country.xlsx"))
region_clean = pd.read_excel(os.path.join(data_path, "Region.xlsx"))
continent_clean = pd.read_excel(os.path.join(data_path, "Continent.xlsx"))
mode_clean = pd.read_excel(os.path.join(data_path, "Mode.xlsx"))
type_clean = pd.read_excel(os.path.join(data_path, "Type.xlsx"))
item_clean = pd.read_excel(os.path.join(data_path, "Item.xlsx"))
transaction_clean = pd.read_excel(os.path.join(data_path, "Transaction.xlsx"))

# Load the additional attraction dataset
updated_item_clean = pd.read_excel(
    os.path.join(data_path, "Additional_Data_for_Attraction_Sites", "Updated_Item.xlsx")
)

print("All datasets loaded successfully.")
```

### Step 2.1: Missing Value Handling

The datasets were checked for missing values during the inspection stage.
A small number of missing values were identified in User and City data.

We will preserve these missing values for now rather than making
incorrect assumptions. They will be handled during the merging and
machine learning preprocessing stages.

```python
# ============================================================
# Check Missing Values in All Datasets
# ============================================================

for name, dataset in datasets.items():
    missing = dataset.isnull().sum()
    missing = missing[missing > 0]

    print(f"\n{name} Dataset:")
    
    if len(missing) == 0:
        print("No missing values.")
    else:
        print(missing)
```

### Step 2.2: Identify Placeholder Values

Some tourism reference datasets use "-" to represent an unknown or
unavailable value. We will identify these placeholders before deciding
how to handle them during preprocessing.

```python
# ============================================================
# Check for "-" Placeholder Values
# ============================================================

text_columns = {
    "City": city_clean,
    "Country": country_clean,
    "Region": region_clean,
    "Continent": continent_clean,
    "Mode": mode_clean,
    "Type": type_clean,
    "Item": item_clean,
    "Updated Item": updated_item_clean
}

for name, dataset in text_columns.items():
    print(f"\n{name} Dataset:")
    
    found = False
    
    for column in dataset.select_dtypes(include="object").columns:
        count = (dataset[column].astype(str).str.strip() == "-").sum()
        
        if count > 0:
            print(f"{column}: {count}")
            found = True
    
    if not found:
        print("No '-' placeholders.")
```

### Step 2.3: Standardize Text Columns

Text columns are standardized by removing leading and trailing spaces.
This helps prevent mismatches when datasets are merged.

# Step 5: Machine Learning — Rating Prediction

## Step 5.1: Prepare Features for Regression

The objective is to predict the tourist Rating using:
- Visit year and month
- Visit mode
- Tourist continent and region
- Attraction
- Attraction type

TransactionId and UserId are excluded because they are identifiers and do not provide meaningful predictive information.

```python
# ============================================================
# Step 5.1: Prepare Features for Rating Prediction
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Select useful features
features = [
    "VisitYear",
    "VisitMonth",
    "VisitMode",
    "UserContinent",
    "UserRegion",
    "Attraction",
    "AttractionType"
]

X = merged_df[features]
y = merged_df["Rating"]

# Identify categorical columns
categorical_features = [
    "VisitMode",
    "UserContinent",
    "UserRegion",
    "Attraction",
    "AttractionType"
]

# Numeric columns
numeric_features = [
    "VisitYear",
    "VisitMonth"
]

print("Features selected:", features)
print("X shape:", X.shape)
print("y shape:", y.shape)
```

### Step 5.2: Train the Rating Prediction Model

The dataset is divided into:
- 80% training data for learning patterns
- 20% testing data for evaluating the model

Categorical features are converted into numerical form using One-Hot Encoding.

A Random Forest Regressor is then trained to predict tourist ratings.

```python
# ============================================================
# Step 5.2: Train Rating Prediction Model
# ============================================================

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Preprocessing:
# One-Hot Encode categorical features
# Keep numerical features unchanged
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", OneHotEncoder(handle_unknown="ignore"),
         categorical_features),
        ("numeric", "passthrough", numeric_features)
    ]
)

# Create Random Forest regression pipeline
regression_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

# Train the model
regression_model.fit(X_train, y_train)

print("Training completed successfully!")
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

### Step 5.3: Evaluate the Rating Prediction Model

The trained model is evaluated on unseen testing data using:
- MAE (Mean Absolute Error)
- MSE (Mean Squared Error)
- R² Score

These metrics help measure how accurately the model predicts tourist ratings.

```python
# ============================================================
# Step 5.3: Evaluate Rating Prediction Model
# ============================================================

# Make predictions on the test data
y_pred = regression_model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Rating Prediction Model Performance")
print("------------------------------------")
print("Mean Absolute Error (MAE):", round(mae, 4))
print("Mean Squared Error (MSE):", round(mse, 4))
print("R² Score:", round(r2, 4))
```

### Step 5.4: Create Historical Rating Features

To improve rating prediction, we create historical features from the training data.

These features include:
- Average rating given by each user
- Average rating received by each attraction
- Number of ratings received by each attraction

Historical rating features are calculated using the actual UserId and Attraction.

Only the training data is used so that information from the test set does not leak into the model.

```python
# ============================================================
# Step 5.4: Correct Historical Rating Features
# ============================================================

# Get the actual UserId values corresponding to training rows
train_data = merged_df.loc[X_train.index, ["UserId", "Attraction"]].copy()
train_data["Rating"] = y_train

# Get the actual UserId values corresponding to testing rows
test_data = merged_df.loc[X_test.index, ["UserId", "Attraction"]].copy()
test_data["Rating"] = y_test

# ------------------------------------------------------------
# Average rating given by each user
# ------------------------------------------------------------
user_avg_rating = (
    train_data.groupby("UserId")["Rating"]
    .mean()
)

# ------------------------------------------------------------
# Average rating received by each attraction
# ------------------------------------------------------------
attraction_avg_rating = (
    train_data.groupby("Attraction")["Rating"]
    .mean()
)

# ------------------------------------------------------------
# Number of ratings received by each attraction
# ------------------------------------------------------------
attraction_rating_count = (
    train_data.groupby("Attraction")["Rating"]
    .count()
)

print("Historical rating features created successfully!")
print("Unique users:", len(user_avg_rating))
print("Unique attractions:", len(attraction_avg_rating))
```

### Step 5.5: Add Historical Rating Features

### Step 7.1: Prepare Attraction Data for Recommendation

A content-based recommendation system recommends attractions based on
their characteristics.

The system uses attraction type, address, and average rating to identify
attractions that may be relevant to a visitor.

```python
# ============================================================
# Step 7.1: Prepare Attraction Data
# ============================================================

# Create attraction-level information
attraction_data = (
    merged_df.groupby(
        [
            "AttractionId",
            "Attraction",
            "AttractionType",
            "AttractionAddress"
        ],
        dropna=False
    )
    .agg(
        AverageRating=("Rating", "mean"),
        RatingCount=("Rating", "count")
    )
    .reset_index()
)

# Sort attractions by average rating
attraction_data = attraction_data.sort_values(
    by="AverageRating",
    ascending=False
).reset_index(drop=True)

print("Attraction recommendation data prepared successfully!")
print("Number of attractions:", len(attraction_data))

# Display the first 10 attractions
attraction_data.head(10)
```

### Step 7.2: Create Content Features

TF-IDF (Term Frequency-Inverse Document Frequency) is used to convert
text-based attraction information into numerical features.

Attraction type and address are combined to calculate similarity between
different attractions.

```python
# ============================================================
# Step 7.2: Create Content Features
# ============================================================

from sklearn.feature_extraction.text import TfidfVectorizer

# Convert missing text values to empty strings
attraction_data["AttractionType"] = (
    attraction_data["AttractionType"].fillna("")
)

attraction_data["AttractionAddress"] = (
    attraction_data["AttractionAddress"].fillna("")
)

# Combine attraction type and address
attraction_data["Content"] = (
    attraction_data["AttractionType"] + " " +
    attraction_data["AttractionAddress"]
)

# Create TF-IDF vectorizer
tfidf = TfidfVectorizer(stop_words="english")

# Convert attraction content into TF-IDF features
tfidf_matrix = tfidf.fit_transform(
    attraction_data["Content"]
)

print("TF-IDF features created successfully!")
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

### Step 7.3: Calculate Attraction Similarity

Cosine similarity is used to measure the similarity between attractions
based on their TF-IDF features.

A higher similarity score means that two attractions have more similar
content characteristics.

```python
# ============================================================
# Step 7.3: Calculate Cosine Similarity
# ============================================================

from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between all attractions
similarity_matrix = cosine_similarity(tfidf_matrix)

print("Cosine similarity calculated successfully!")
print("Similarity matrix shape:", similarity_matrix.shape)
```

### Step 7.4: Build the Recommendation Function

A recommendation function is created to suggest similar attractions
based on a selected attraction.

The system returns the top recommended attractions along with their
average ratings.

```python
# ============================================================
# Step 7.4: Build Recommendation Function
# ============================================================

def recommend_attractions(attraction_name, top_n=5):
    """
    Recommend attractions similar to the selected attraction.
    """

    # Find the index of the selected attraction
    matching_indices = attraction_data[
        attraction_data["Attraction"].str.lower() == attraction_name.lower()
    ].index

    # Check whether the attraction exists
    if len(matching_indices) == 0:
        return "Attraction not found."

    # Get the attraction index
    idx = matching_indices[0]

    # Get similarity scores for the selected attraction
    similarity_scores = list(enumerate(similarity_matrix[idx]))

    # Sort attractions by similarity score
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Remove the selected attraction itself
    similarity_scores = [
        item for item in similarity_scores
        if item[0] != idx
    ]

    # Select top recommendations
    top_indices = [item[0] for item in similarity_scores[:top_n]]

    # Return recommendation details
    recommendations = attraction_data.iloc[top_indices][
        [
            "Attraction",
            "AttractionType",
            "AverageRating",
            "RatingCount"
        ]
    ].copy()

    recommendations["AverageRating"] = recommendations[
        "AverageRating"
    ].round(2)

    return recommendations.reset_index(drop=True)


# Test the recommendation system
recommendations = recommend_attractions(
    "Waterbom Bali",
    top_n=5
)

recommendations
```

### Step 7.5: Improve Recommendation Ranking

The recommendation system combines content similarity with attraction
ratings.

This helps prioritize attractions that are both similar to the selected
attraction and highly rated by previous visitors.

```python
# ============================================================
# Step 7.5: Improved Recommendation Ranking
# ============================================================

def recommend_attractions(attraction_name, top_n=5):
    """
    Recommend attractions using content similarity and average rating.
    """

    # Find the selected attraction
    matching_indices = attraction_data[
        attraction_data["Attraction"].str.lower() == attraction_name.lower()
    ].index

    if len(matching_indices) == 0:
        return "Attraction not found."

    idx = matching_indices[0]

    # Get similarity scores
    similarity_scores = list(enumerate(similarity_matrix[idx]))

    # Remove the selected attraction
    similarity_scores = [
        item for item in similarity_scores
        if item[0] != idx
    ]

    # Create recommendation dataframe
    recommendation_indices = [
        item[0] for item in similarity_scores
    ]

    recommendations = attraction_data.iloc[
        recommendation_indices
    ].copy()

    # Add similarity score
    score_lookup = dict(similarity_scores)

    recommendations["SimilarityScore"] = recommendations.index.map(
        score_lookup
    )

    # Normalize rating from 0 to 1
    recommendations["RatingScore"] = (
        recommendations["AverageRating"] / 5
    )

    # Combined recommendation score
    recommendations["RecommendationScore"] = (
        0.7 * recommendations["SimilarityScore"] +
        0.3 * recommendations["RatingScore"]
    )

    # Sort by combined score
    recommendations = recommendations.sort_values(
        "RecommendationScore",
        ascending=False
    )

    # Return top recommendations
    recommendations = recommendations.head(top_n)

    return recommendations[
        [
            "Attraction",
            "AttractionType",
            "AverageRating",
            "RatingCount",
            "SimilarityScore",
            "RecommendationScore"
        ]
    ].reset_index(drop=True)


# Test improved recommendation system
recommendations = recommend_attractions(
    "Waterbom Bali",
    top_n=5
)

recommendations
```
